```python
import pandas as pd
import random
from IPython.display import Image, display

# Loading the watchlist csv file
df = pd.read_csv("Watchlist - Sheet1.csv")
df["Genre"]=df["Genre"].str.strip().str.lower()

# Start Loop
while True:
    filter_genre = input("\n🎭Enter a genre to filter (or leave blank for any, or type 'quit' to quit): ").strip().lower()

    if filter_genre == "exit":
        print("👋 Goodbye! Enjoy your binge session!")
    
        break

# Apply filters
    filtered_df = df.copy()
    if filter_genre:
        filtered_df = filtered_df[filtered_df["Genre"] == filter_genre]


# Choose Randomly
    if not filtered_df.empty:
        choice = filtered_df.sample(1).iloc[0]
        print(f"\n🎬 Let's Watch: {choice['Title']} ({choice['Type']}) on {choice['Platform']}")

    # Display the image
        if "Image" in choice and pd.notnull(choice["Image"]):
            display(Image(url=choice["Image"], width=250))
    else:
      print("❌ No matches found. Try another genre!")
```
